Import all Libraries required for analysis

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

Now enter data from your dataset

```python
df = pd.read_csv('/content/Books_Data_Clean.csv')
```

```python
df.duplicated().sum()
```

To check unique values

```python
df.nunique()
```

Creating a histogram of number of books published in that year

```python
plt.hist(df["Publishing Year"])
plt.xlabel("Publishing Year")
plt.ylabel("Frequency")
plt.title("Distribution of Publishing Year")
plt.show()
```

.value_counts() function returns object containing counts of unique values.

```python
df["genre"].value_counts()
```

```python
df["genre"].value_counts().plot(kind="bar")
plt.xlabel("Genre")
plt.ylabel("Frequency")
plt.title("Distribution of Genre")
plt.show()
```

```python
df.groupby("Author")["Book_average_rating"].mean()
```

```python
df.groupby("Author")["Book_average_rating"].mean().sort_values(ascending=False)
```

```python
sns.boxplot(x = "genre" , y = "Book_average_rating" , data = df)
plt.xlabel("Genre")
plt.ylabel("Book_average_rating")
plt.title("Boxplot of Genre vs Book_average_rating")
plt.show()
```

```python
plt.scatter(df["sale price"], df["units sold"])
plt.xlabel("Sale Price")
plt.ylabel("Units Sold")
plt.title("Scatter Plot of Sale Price vs Units Sold")
plt.show()
```

```python
language_count = df["language_code"].value_counts()
```

```python
plt.pie(language_count, labels = language_count.index, autopct = "%1.1f%%")
plt.title("Distribution of Languages")
plt.show()
```

```python
df.groupby('Publisher ')["publisher revenue"].sum().sort_values(ascending = False)
```

```python
df.groupby("Author_Rating")["Book_ratings_count"].mean().sort_values(ascending = False)
```

```python
plt.scatter(df["Book_average_rating"], df['Book_ratings_count'])
plt.xlabel("Book_average_rating")
plt.ylabel("Book_ratings_count")
plt.title("Scatter Plot of Author_Rating vs Book_ratings_count")
plt.show()
```

```python
total_gross_sales = df.groupby("Author")["gross sales"].sum()
```

```python
total_gross_sales.sort_values(ascending = False).head(20).plot(kind = "bar")
plt.xlabel("Author")
plt.ylabel("Total Gross Sales")
plt.title("Top 10 Authors by Total Gross Sales")
plt.show()
```

```python
df.groupby("Publishing Year")["units sold"].sum().plot(kind ="line")
plt.xlabel("Publishing Year")
plt.ylabel("Units Sold")
plt.title("Units Sold Over Time")
plt.show()
```

```python
sns.boxplot(x= "Author_Rating", y = "units sold", data = df)
plt.xlabel("Author_Rating")
plt.ylabel("units sold")
plt.title("Boxplot of Author_Rating vs units sold")
plt.show()
```
